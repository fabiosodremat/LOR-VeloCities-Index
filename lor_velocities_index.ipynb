## L’OR VeloCities Index: “A Mathematical Model of Urban Rhythm.”


## 1. Introduction

This study was developed by Fábio Sodré Rocha, PhD in Mathematics,  in collaboration with the coffee brand L’OR.

The perception that some cities operate at a faster pace than others is universal. This work proposes a quantitative framework to measure such differences, translating urban tempo into a scalar index.

The proposed model is based on observable variables and mathematical transformations designed to produce an interpretable and dimensionless index.

The central objective is to construct a dimensionless index capable of translating complex urban dynamics into a scalar quantity interpretable on a perceptual scale analogous to media playback speed (e.g., 0.75x, 1.0x, 2.0x).

The model is based exclusively on observable variables and combines elements of:

- urban dynamics  
- statistical normalization  
- perceptual scaling  

## 1. Fundamental Variables

Let $i$ index a given city.

We define:

$
D_i \in \mathbb{R}^+ \quad \text{(population density, inhabitants/km²)}
$

$
T_i \in \mathbb{R}^+ \quad \text{(average one-way commute time, minutes)}
$

These variables encode two orthogonal dimensions of urban experience:

- **spatial compression** (density of agents per unit area)
- **temporal cost** (time required for displacement)

Together, they form a minimal yet expressive representation of urban intensity.


## 2. Definition of the Urban Load Functional

We define the Urban Load Index:

$$
URI_i = D_i \cdot T_i
$$

This quantity can be interpreted as a first-order proxy for **urban friction**, representing the interaction between spatial congestion and temporal expenditure.

From a modeling perspective:

- $D_i$ captures pressure on space
- $T_i$ captures pressure on time
- their product captures **joint systemic stress**

Thus:

$$
URI_i \propto \text{urban pressure}
$$

## 3. Nonlinear Transformation

We introduce the transformation:

$$
X_i = \sqrt{URI_i}
$$

### Justification

The raw scale of $URI_i$ is highly dispersed and exhibits heavy-tailed behavior.

The square root transformation serves to:

- reduce variance
- compress extreme values
- approximate perceptual scaling laws

This aligns with classical psychophysical principles, where perceived intensity often follows sublinear transformations of physical stimuli.

## 4. Geometric Normalization

We define the geometric mean:

$$
\bar{X}_g = \exp\left(\frac{1}{n} \sum_{i=1}^{n} \ln X_i \right)
$$

and the normalized index:

$$
R_i^{(base)} = \frac{X_i}{\bar{X}_g}
$$


The geometric mean is preferred over the arithmetic mean due to:

- the multiplicative structure of the model
- the log-normal tendency of the data
- robustness to extreme values

This normalization renders the index **dimensionless** and relative.

## 5. Rescaling

We define the normalized index as:

$$
R_i = 2 \cdot \frac{R_i^{(base)}}{R_{\max}}
$$

where:

$
R_{\max} = \max_i R_i^{(base)}
$

### Properties

This normalization ensures that:

- the maximum value is mapped to $2.0$,
- all values remain strictly positive,
- proportional relationships between observations are preserved.

The resulting index is dimensionless and scale-consistent.

## 6. Interpretation

The index satisfies:

$$
0 < R_i \leq 2.0
$$

Interpretation:

- $R \approx 2.0$: highly accelerated urban environment
- $R \approx 1.0$: moderate rhythm
- smaller values: slower, more contemplative environments

All values retain meaningful magnitude without artificial truncation.

## 7. Final Expression

The complete model is given by:

$$
R_i =
2 \cdot
\frac{
\frac{\sqrt{D_i \cdot T_i}}{\bar{X}_g}
}{
R_{\max}
}
$$

where:

$$
\bar{X}_g = \exp\left(\frac{1}{n} \sum_{j=1}^{n} \ln \sqrt{D_j \cdot T_j} \right)
$$

and:

$$
R_{\max} = \max_j \left( \frac{\sqrt{D_j \cdot T_j}}{\bar{X}_g} \right)
$$

This defines a dimensionless and strictly positive measure of urban tempo.

## 8. Data

The dataset used in this study was collected and provided by Ma. Adriana Alves.

It includes population density and average one-way commute time for a set of selected cities.

The following code cell performs the computational steps required to construct the Urban Rhythm Index based on this dataset.

```python
import numpy as np
import pandas as pd


# Data
data = [
    ("Mexico City", 9017, 67),
    ("New York City", 11313, 51),
    ("Sao Paulo", 7528, 55),
    ("Rio de Janeiro", 5220, 58),
    ("Shanghai", 5436, 54),
    ("London", 5640, 50),
    ("Brussels", 7500, 37),
    ("Tokyo", 6402, 43),
    ("Paris", 5340, 47),
    ("Berlin", 4200, 36),
    ("Amsterdam", 5277, 28),
   # ("Porto Alegre", 2653, 48),
    ("Capri", 1250, 15),
    ("Santorini", 170, 22),
    ("Oslo", 1699, 26)
]

df = pd.DataFrame(data, columns=["City", "D", "T"])

# Step 1: URI
df["URI"] = df["D"] * df["T"]

# Step 2: Transformation
df["X"] = np.sqrt(df["URI"])

# Step 3: Geometric Mean
Xg = np.exp(np.mean(np.log(df["X"])))

# Step 4: Base Index
df["R_base"] = df["X"] / Xg

# Step 5: Rescaling
R_max = df["R_base"].max()
df["R"] = 2 * (df["R_base"] / R_max)
df["R"] = df["R"].round(2)

# Step 6: Formatting
df["Ritmo"] = df["R"].astype(str) + "x"

# Output
result = df[["City", "D", "T", "URI", "R", "Ritmo"]].sort_values("R", ascending=False)

print(result)
```

## 9. Scatter Plot: R vs URI

The following plot illustrates the relationship between the Urban Load Index ($URI$) and the Urban Rhythm Index ($R$).

It allows us to visualize how the transformation from raw urban intensity to normalized rhythm behaves across cities.

```python
import matplotlib.pyplot as plt

# Scatter plot
plt.figure()
plt.scatter(df["URI"], df["R"])

# Labels
plt.xlabel("URI (D × T)")
plt.ylabel("L'Or Index (R)")
plt.title("Scatter Plot: R vs URI")

# Opcional: adicionar nomes das cidades
for i, city in enumerate(df["City"]):
    plt.text(df["URI"][i], df["R"][i], city, fontsize=8)

plt.show()
```

##10. Interpretation of the Scatter Plot: R vs URI

The scatter plot shows the relationship between the Urban Load Index ($URI = D \cdot T$) and the Urban Rhythm Index ($R$).

The relationship is monotonic: higher $URI$ values correspond to higher $R$ values, preserving the ranking of cities.

However, the relationship is nonlinear due to the square root transformation:

$$
X_i = \sqrt{URI_i}
$$

This produces a sublinear effect, where increases in $URI$ lead to diminishing increases in $R$.

As a result, extreme values are compressed, and the index becomes more stable and interpretable.

In summary, $URI$ measures raw urban intensity, while $R$ captures a normalized and perceptually adjusted version of this intensity.

Through a transformation pipeline involving nonlinear scaling and geometric normalization, the model preserves the relative structure of the data while improving interpretability and reducing the influence of extreme values.

The resulting index $R$ provides a dimensionless and scalable representation of urban intensity, enabling meaningful comparisons across heterogeneous environments.
